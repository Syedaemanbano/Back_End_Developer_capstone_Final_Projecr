```python
import yaml

# Define the new configuration
data_config = {
    'train': '/content/drive/MyDrive/Military Vehicles Dataset.v1i.yolo26/train/images',
    'val': '/content/drive/MyDrive/Military Vehicles Dataset.v1i.yolo26/valid/images',
    'test': '/content/drive/MyDrive/Military Vehicles Dataset.v1i.yolo26/test/images',
    'nc': 8,
    'names': ['APC', 'Tanks-v5', 'car', 'military-vehicle', 'soldier', 'tank', 'vehicle', 'zirhli-arac']
}

# Write it to your drive
yaml_path = '/content/drive/MyDrive/Military Vehicles Dataset.v1i.yolo26/data.yaml'
with open(yaml_path, 'w') as f:
    yaml.dump(data_config, f)

print("data.yaml has been updated for Colab!")
```

```python
from ultralytics import YOLO

# 1. Load the YOLO26 Nano model
# (Nano is best for real-time video tracking)
model = YOLO("yolo26n.pt")

# 2. Start training
model.train(
    data="/content/drive/MyDrive/Military Vehicles Dataset.v1i.yolo26/data.yaml",
    epochs=100,
    imgsz=640,
    batch=16,         # Adjust this if you get a 'CUDA out of memory' error
    project="/content/drive/MyDrive/Military_Project_Results",
    name="vehicle_detection_run",
    device=0          # Force the use of the T4 GPU
)
```

```python
import os
weight_path = "/content/drive/MyDrive/Military_Project_Results/vehicle_detection_run/weights/best.pt"
if os.path.exists(weight_path):
    print("Found best.pt! You are ready to test.")
else:
    print("Weights not found. Check your folder path.")
```
